```python
from datetime import datetime, date, time
print('<디데이(D-day) 프로그램')
print( )
year=int(input('연도를 입력하세요: '))
month=int(input('월을 입력하세요: '))
day=int(input('일을 입력하세요: '))
dt4=date(year,month,day)
today=date.today()
dday=(dt4-today).days
print(f'[{year}년 {month}월 {day}일은 {dday}일 남았습니다.')
```
